# 数据切分

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
X=data.drop(columns=['tem-dif'])
y=data['tem-dif']
X_train,X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=202)
print(X_train.shape,X_test.shape)
```

# 定义评价函数

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

def try_different_method(y_pred_train1, y_pred_test1, y_train, y_test):
    print('训练集:')
    
    r2 = r2_score(y_train, y_pred_train1)
    mae = mean_absolute_error(y_train, y_pred_train1)
    mse = mean_squared_error(y_train, y_pred_train1)
    rmse = np.sqrt(mse)  # 手动计算 RMSE
    
    p1 = (y_pred_train1 / y_train).mean()
    print("R2:     ", '%.4f' % float(r2),
          "绝对误差: ", '%.4f' % float(mae),
          "均方误差: ", '%.4f' % float(mse),
          "均方根误差:", '%.4f' % float(rmse))
    
    train_metrics = [r2, mae, mse, rmse]
    print('*' * 50)
    print('测试集:')
    
    r2 = r2_score(y_test, y_pred_test1)
    mae = mean_absolute_error(y_test, y_pred_test1)
    mse = mean_squared_error(y_test, y_pred_test1)
    rmse = np.sqrt(mse)
    
    p1 = (y_pred_test1 / y_test).mean()
    print("R2:     ", '%.4f' % float(r2),
          "绝对误差: ", '%.4f' % float(mae),
          "均方误差: ", '%.4f' % float(mse),
          "均方根误差:", '%.4f' % float(rmse))
    
    test_metrics = [r2, mae, mse, rmse]
    return train_metrics, test_metrics
```

## 调参

```python
import optuna
import warnings
import xgboost as xgb
def optuna_opti(trial):
    # 定义待优化的参数及其搜索范围
    params={'learning_rate': trial.suggest_loguniform('learning_rate',1e-3,0.5), 
            'max_depth': trial.suggest_int('max_depth',2,31),
            'reg_alpha': trial.suggest_loguniform('reg_alpha',1e-1,100), 
            'reg_lambda': trial.suggest_loguniform('reg_lambda',1e-1,100), 
            # 'min_child_samples': trial.suggest_int('min_child_samples',2,5),
            'n_estimators': trial.suggest_int('n_estimators',1,500),
            'colsample_bytree':trial.suggest_loguniform('colsample_bytree',0.5,1), 
            'subsample':trial.suggest_loguniform('subsample',0.5,1), 
#              'class_weight':'balanced',
            'random_state':1,
            'n_jobs':-1}
    clf=xgb.XGBRegressor(**params)
    clf.fit(X_train,y_train)
    score=clf.score(X_test,y_test)
    return score
#实际优化过程，首先实例化优化器
study = optuna.create_study(direction="maximize" #优化的方向，可以填写minimize或maximize
                           )
#开始优化，n_trials为允许的最大迭代次数
#由于参数空间已经在目标函数中定义好，因此不需要输入参数空间
study.optimize(optuna_opti #目标函数
               , n_trials=30 #最大迭代次数（包括最初的观测值的）
               , show_progress_bar=True #展示进度条
              )

#可直接从优化好的对象study中调用优化的结果
#打印最佳参数与最佳损失值
print("best params: ", study.best_trial.params,"best score: ", study.best_trial.values)      
```

# Xgboost

```python
import xgboost as xgb
params=study.best_trial.params
clf=xgb.XGBRegressor(**params,random_state=1,n_jobs=-1)
clf.fit(X_train,y_train)
y_pred_train2=clf.predict(X_train)
y_pred_test2=clf.predict(X_test)
print('模型评分如下：\n')
train_metrics,test_metrics=try_different_method(y_pred_train2,y_pred_test2,y_train,y_test)
```

```python
# 创建主图（用来画蜂巢图）
fig, ax1 = plt.subplots(figsize=(8, 4), dpi=120)
shap.summary_plot(np.array(shap_values2),X_test, feature_names=X_test.columns, plot_type="dot", show=False, color_bar=True)
plt.gca().set_position([0.2, 0.2, 0.65, 0.65])  # 调整图表位置，留出右侧空间放热度条
# 获取共享的 y 轴
ax1 = plt.gca()
# 创建共享 y 轴的另一个图，绘制特征贡献图在顶部x轴
ax2 = ax1.twiny()
shap.summary_plot(np.array(shap_values2),X_test,  feature_names=X_test.columns, plot_type="bar", show=False)
plt.gca().set_position([0.2, 0.2, 0.65, 0.65])  # 调整图表位置，与蜂巢图对齐
# 在顶部 X 轴添加一条横线
# ax2.axhline(y=8, color='gray', linestyle='-', linewidth=1)  # 注意y值应该对应顶部
# 调整透明度
bars = ax2.patches  # 获取所有的柱状图对象
for bar in bars:
    bar.set_alpha(0.2)  # 设置透明度
# 设置两个x轴的标签
ax1.set_xlabel('Shap Value (impact on model output)', fontsize=12)
ax2.set_xlabel('Mean(|SHAP value|) (Feature Importance)', fontsize=12)
# 移动顶部的 X 轴，避免与底部 X 轴重叠
fig.set_size_inches(8, 4)  # 覆盖之前的尺寸设置
ax2.xaxis.set_label_position('top')  # 将标签移动到顶部
ax2.xaxis.tick_top()#将刻度也移动到顶部
ax2.spines['top'].set_visible(True)
ax2.spines['top'].set_linewidth(1.2)
ax2.spines['top'].set_color('black')
# 设置y轴标签
# ax1.set_ylabel('Feature', fontsize=12)
plt.tight_layout() 
plt.savefig("SHAP-组合图.jpg",dpi=600, bbox_inches = 'tight')
plt.show()
```

```python
import shap
import matplotlib.pyplot as plt
import numpy as np

# 计算特征重要性（按均值绝对 SHAP 值排序）
importance = np.abs(shap_values1.values).mean(axis=0)
features = X_test.columns
sorted_idx = np.argsort(importance)  # 排序索引
sorted_features = features[sorted_idx]  # 排序后的特征
sorted_importance = importance[sorted_idx]  # 排序后的重要性

# 🎨 生成颜色映射（所有变量颜色统一）
colors = plt.cm.viridis(np.linspace(0, 1, len(sorted_features)))

# 创建主图（SHAP 重要性条形图）
fig, ax1 = plt.subplots(figsize=(10, 6),dpi=120)

# 绘制条形图，颜色按照变量匹配
ax1.barh(sorted_features, sorted_importance, color=colors, alpha=0.8)
ax1.set_xlabel("Mean(|SHAP value|) (Feature Importance)")
ax1.set_title("SHAP Feature Importance")
# ✨ **添加数值标签**
for i, v in enumerate(sorted_importance):
    ax1.text(v + 0.002, i, f"{v:.2f}", va="center", fontsize=10, color="black")
# 🎨 饼图颜色也保持一致（选取最后 5 个重要变量的颜色）
num_feature=8
selected_features = sorted_features[-num_feature:]  # 取最重要的 5 个变量
selected_importance = sorted_importance[-num_feature:]  # 取对应的重要性
selected_colors = colors[-num_feature:]  # 颜色匹配

# 在右下角创建嵌套坐标
ax2 = fig.add_axes([0.4, 0.2, 0.5, 0.5])  # [左, 下, 宽, 高]

# 画 2D 饼图（手动加阴影，模拟 3D 效果）
explode = [0.1] * len(selected_features)  # 让所有扇形稍微分开

ax2.pie(
    selected_importance / selected_importance.sum(),  # 计算占比
    labels=selected_features, 
    autopct="%1.1f%%", 
    colors=selected_colors,  # 颜色匹配条形图
    startangle=140, 
    explode=explode,
    shadow=False  # 添加阴影，制造 3D 效果
)

# ax2.set_title("Feature Importance Distribution")
plt.savefig('importance.jpg',dpi=600, bbox_inches = 'tight')
# 调整布局
plt.show()
```

```python
#依赖图画前8个
list11=pd.DataFrame({'shap':abs(shap_values2).mean(axis=0),'fea':feature_names}).sort_values(by='shap',ascending=False).head(8).fea.tolist()
```

```python
fig, axes = plt.subplots(7, 4, figsize=(15, 20),dpi=120)
axes = axes.flatten()
ii=1
for feature_i in range(len(list11)):
    for feature_j in range(len(list11)):
        if feature_i<feature_j:
            ax = axes[ii - 1]  
            # 提取交互值矩阵中(i,j)的列向量
            interaction_values = shap_interaction_values[:, feature_i, feature_j]
            # 构造二维SHAP值矩阵：将交互值填充到第i列，其他列设为0
            shap_values_2d = np.zeros((X_test.shape[0], len(feature_names)))
            shap_values_2d[:, feature_i] = interaction_values
            shap.dependence_plot(
                ind=feature_i,                # 主特征（x轴）
                shap_values=shap_values_2d,   # 二维矩阵，第i列为交互值
                features=X_test,                   # 原始数据
                interaction_index=feature_j,  # 颜色编码的交互特征j
                show=False,
                ax=ax 
            )
            ax.axhline(y = 0, color='black', linestyle='-.', linewidth=1, label='SHAP value = 0')
            # plt.axvline(x = 350, color='black', linestyle='-.', linewidth=1, label='SHAP value = 0')
            ax.set_ylabel("SHAP Interaction Value")
            ii += 1
plt.tight_layout()
plt.savefig("SHAP Interaction Value.jpg",dpi=600, bbox_inches = 'tight')
plt.show()
```

```python
from statsmodels.nonparametric.smoothers_lowess import lowess
from scipy.optimize import fsolve
ii = 1
fig, axes = plt.subplots(2, 4, figsize=(15, 8),dpi=120)##2行4列。8个子图。如果变量数量太多，需要修改。
axes = axes.flatten() 
for col in list11:
    ax = axes[ii - 1]
    index=feature_names.index(col)
    x1=X_test.iloc[:,index]
    x2=shap_values2[:,index]
    ax.scatter(x1, x2, color='dodgerblue', s = 6)
    # 手动计算LOWESS拟合，设置带宽（例如带宽为0.3，你可以根据实际情况调整）
    lowess_fit = lowess(x2, x1, frac=0.3)
    x_fit = lowess_fit[:, 0]
    y_fit = lowess_fit[:, 1]
    ax.plot(x_fit, y_fit, color='red')

    # 找出所有与y = 0相交的x值
    def find_zero_crossings(x_fit, y_fit):
        crossings = []
        for i in range(1, len(y_fit)):
            if (y_fit[i - 1] < 0 and y_fit[i] > 0) or (y_fit[i - 1] > 0 and y_fit[i] < 0):
                # 使用插值法找到x_fit和y_fit中y值接近0的x值
                crossing = fsolve(lambda x: np.interp(x, x_fit, y_fit), x_fit[i])[0]
                crossings.append(crossing)
        return crossings

    x_intercepts = find_zero_crossings(x_fit, y_fit)
    # 在图中标注所有的x_intercepts
    for x_intercept in x_intercepts:
        if x_intercept>0:
            ax.axvline(x=x_intercept, color='blue', linestyle='--', label=f'Intersection at Value = {x_intercept:.2f}')
    #    plt.text(x_intercept, 0.2, f'Value = {x_intercept:.2f}', color='blue', fontsize=10, verticalalignment='bottom')
    # 添加shap = 0的横线
    ax.axhline(y = 0, color='black', linestyle='-.', linewidth=1)
    # 添加图例
    ax.legend(fontsize=10, frameon=False)
    # 设置标签和标题
    ax.set_xlabel(f'{col}')
    ax.set_ylabel('shap values')
    # 去除上和右的边框线
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # 显示图表
    ii += 1
    # 隐藏多余的空子图
    for j in range(len(list11), len(axes)):
        fig.delaxes(axes[j])
plt.tight_layout()
plt.savefig('dependence_plot1.jpg',dpi=600, bbox_inches = 'tight')
plt.show()
```

```python
ii = 1
fig, axes = plt.subplots(2, 4, figsize=(15, 8),dpi=120)##2行4列。8个子图。如果变量数量太多，需要修改。
axes = axes.flatten() 
for i in range(len(list11)):
    ax = axes[ii - 1]  
    shap.dependence_plot(
        list11[i], shap_values2, X_test, feature_names=feature_names,
        interaction_index=None, show=False, 
        ax=ax  
    )
    x1=X_test[list11[i]]
    x2=pd.DataFrame(shap_values2,columns=feature_names)[list11[i]]
    lowess_fit = lowess(x2, x1, frac=0.3)
    x_fit = lowess_fit[:, 0]
    y_fit = lowess_fit[:, 1]
    ax.plot(x_fit, y_fit, color='red', label='LOWESS Curve')
    
#     ax.set_title(f'{feature_names[i]}')
    ii += 1
    # 隐藏多余的空子图
for j in range(len(feature_names), len(axes)):
    fig.delaxes(axes[j])
plt.tight_layout()
plt.savefig('dependence_plot3.jpg',dpi=600, bbox_inches = 'tight')
plt.show()
```

```python
import shap
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

images = []
for i, col in enumerate(list11):
    fig_tmp, ax_tmp = shap.plots.partial_dependence(
        col, clf.predict, X_test,
        ice=False,
        model_expected_value=True,
        feature_expected_value=True,
        show=False
    )
    fname = f"tmp_{col}.png"
    fig_tmp.savefig(fname, dpi=600, bbox_inches='tight')
    images.append(Image.open(fname))
    plt.close(fig_tmp)
# 拼成一张大图
rows, cols = 2, 4
w, h = images[0].size
merged = Image.new('RGB', (cols*w, rows*h), (255, 255, 255))
for i, img in enumerate(images):
    x = (i % cols) * w
    y = (i // cols) * h
    merged.paste(img, (x, y))

merged.save('dependence_plot2.jpg', quality=95)
# merged.show()
# display(merged)
```
